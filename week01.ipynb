```python
import tiktoken
```

```python
encoding = tiktoken.get_encoding("o200k_base")
text = "안녕하세요. 토큰화를 연습합니다."

token_ids = encoding.encode(text)
token_ids
```

- 텍스트가 정수 ID의 나열로 변환
- 각 ID가 o200k_base 어휘 집합의 토큰 하나를 가리킴

```python
for token_id in token_ids:
    token = encoding.decode_single_token_bytes(token_id)
    print(token.decode("utf-8", errors="replace"))
```

- 각 ID에 대응하는 바이트를 UTF-8 문자열로 바꾸어 토큰 경계를 확인
- ' 토'와 ' 연'처럼 앞의 공백이 글자와 함께 묶일 수 있음. 안녕하세요는 '안'과 '녕하세요'로, 연습은 ' 연'과 '습'으로 나뉨
---
#### 한 토큰이 완전한 클자가 아닐 수 있음
- BPE 토큰은 UTF-8 문자의 일부 바이트만 포함할 수도 있음.
- 이 경우 토큰 하나를 문자열로 바꾸면 �가 나타날 수 있음.
- 원문을 복원할 때는 개별 토큰이 아니라 전체 ID 나열을 decode()에 전달.

```python
encoding.decode(token_ids)
```

```python
len(text)  # 문자 수
```

```python
len(token_ids)  # 토큰 수
```

- 문자 수와 토큰 수가 일치하지 않음. LLM이 글자 수 세기 같은 작업을 어려워할 수 있는 이유 중 하나.

```python
english_text = "Hello. We practice tokenization."

english_token_ids = encoding.encode(english_text)
english_token_ids
```

```python
for token_id in english_token_ids:
    token = encoding.decode_single_token_bytes(token_id)
    print(token.decode("utf-8", errors="replace"))
```

```python
len(english_text)  # 영어 문자 수
```

```python
len(english_token_ids)  # 영어 토큰 수
```

- 영어 문장은 32자이지만 7개 토큰, 같은 내용의 한국어 문장은 18자이지만 10개 토큰
- 영어의 자주 쓰이는 문자열이 더 큰 토큰으로 묶여 영어가 더 적은 토큰으로 표현됨.
